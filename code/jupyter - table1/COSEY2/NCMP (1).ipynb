```r
library(bigrquery)
library(knitr)
library(tidyverse)
library(ggplot2)
library(dplyr)
library("table1")
library("IRdisplay")
bq_auth()
```

```r
data <-bq_dataset_query(query="
SELECT *

 FROM `yhcr-prd-phm-bia-core.CB_1935_AK.src_bmbc_NCMP1`"
                       ,x="yhcr-prd-phm-bia-core.CB_1935_AK")

NCMP <- bq_table_download(data) 
```

```r
data1 <-bq_dataset_query(query="
select round(avg(count1),5)

from(

SELECT person_id,count(person_id) count1

 FROM `yhcr-prd-phm-bia-core.CB_1935_AK.src_bmbc_NCMP1`

 group by person_id)base"
                       ,x="yhcr-prd-phm-bia-core.CB_1935_AK")

NCMPAVG <- bq_table_download(data1) 
```

```r
NCMPR<-  unique(filter(NCMP,ageCat == "Reception Age"))
NCMP6<-  unique(filter(NCMP,ageCat == "Year 6 Age"))
```

```r
display_jupyter <- function(x) {
  css <- system.file("table1_defaults_1.0/table1_defaults.css", package="table1")
  css <- paste(readLines(css), collapse="\n")
  x <- htmltools::tagList(htmltools::tags$style(css), htmltools::tags$div(class="Rtable1", x))
  IRdisplay::display_html(as.character(x))
}
```

```r
label(NCMPR$ageCat)      <- "School Year @ time of measure"
label(NCMPR$DateOfMeasure)  <- "Year Of Measure"
label(NCMPR$AvgIMD)      <- "IMD"
label(NCMPR$ethnic_group)      <- "Ethnic Group"
label(NCMPR$gender_source_value)      <- "Gender"
label(NCMPR$BMI)      <- "Body Mass Index (kg/m^2)"

label(NCMP6$ageCat)      <- "School Year @ time of measure"
label(NCMP6$DateOfMeasure)  <- "Year Of Measure"
label(NCMP6$AvgIMD)      <- "IMD"
label(NCMP6$ethnic_group)      <- "Ethnic Group"
label(NCMP6$gender_source_value)      <- "Gender"
label(NCMP6$BMI)      <- "Body Mass Index (kg/m^2)"
```

```r
 xR<-table1(~ NCMPR$DateOfMeasure + NCMPR$AvgIMD+ NCMPR$gender_source_value + NCMPR$ethnic_group|NCMPR$BMI ,data=NCMPR
          ,justify=c("left", "left", rep("center", 5)),format_number = TRUE
          ,caption = "COSEY2: Physical health & health behaviors - 1.2 Reception aged children BMI "
    
         )
```

```r
 x6<-table1(~ NCMP6$DateOfMeasure + NCMP6$AvgIMD+ NCMP6$gender_source_value + NCMP6$ethnic_group|NCMP6$BMI ,data=NCMP6
          ,justify=c("left", "left", rep("center", 5)),format_number = TRUE
          ,caption = "COSEY2: Physical health & health behaviors - 1.2 Year 6 children BMI "
    
         )
```

```r
display_jupyter(xR)
```
